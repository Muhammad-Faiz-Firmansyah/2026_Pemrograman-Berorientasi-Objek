## Praktikum 03: Mendesain dan Membuat Kelas OOP
Hierarki:
- Lokasi (ABC) -> TempatWisata, Kuliner, TempatIbadah, Landmark

```python
from abc import ABC, abstractmethod
class Lokasi(ABC):
    def __init__(self, nama, latitude, longitude):
        self.nama = str(nama) if nama else 'Tanpa Nama'
        try:
            self.latitude = float(latitude)
            self.longitude = float(longitude)
        except (ValueError, TypeError):
            self.latitude = None
            self.longitude = None
            print(f'warning: koordinat {nama} gak valid')
    def get_koordinat(self): return (self.latitude, self.longitude)
    def __repr__(self):
        return f"{self.__class__.__name__}(nama='{self.nama}', lat={self.latitude}, lon={self.longitude})"
    def __str__(self): return f'{self.nama} [{self.__class__.__name__}]'
    @abstractmethod
    def get_info_popup(self): pass
class TempatWisata(Lokasi):
    def __init__(self, nama, lat, lon, desk=''):
        super().__init__(nama, lat, lon)
        self.deskripsi = str(desk) if desk else 'Tempat wisata'
    def get_info_popup(self):
        return f'<h4>{self.nama}</h4><i>Tempat Wisata</i><br>{self.deskripsi}'
class Kuliner(Lokasi):
    def __init__(self, nama, lat, lon, desk=''):
        super().__init__(nama, lat, lon)
        self.deskripsi = str(desk) if desk else 'Tempat kuliner'
    def get_info_popup(self):
        return f'<h4>{self.nama}</h4><i>Kuliner</i><br>{self.deskripsi}'
class TempatIbadah(Lokasi):
    def __init__(self, nama, lat, lon, desk=''):
        super().__init__(nama, lat, lon)
        self.deskripsi = str(desk) if desk else 'Tempat ibadah'
    def get_info_popup(self):
        return f'<h4>{self.nama}</h4><i>Tempat Ibadah</i><br>{self.deskripsi}'
class Landmark(Lokasi):
    def __init__(self, nama, lat, lon, desk=''):
        super().__init__(nama, lat, lon)
        self.deskripsi = str(desk) if desk else 'Landmark'
    def get_info_popup(self):
        return f'<h4>{self.nama}</h4><i>Landmark</i><br>{self.deskripsi}'
# Test
for o in [TempatWisata('A',-6.9,110.4,'test'),Kuliner('B',-6.9,110.4,'test'),
          TempatIbadah('C',-6.9,110.4,'test'),Landmark('D',-6.9,110.4,'test')]:
    print(o)
```

## Praktikum 04: Membuat Objek dari Data Pandas

```python
def buat_objek_lokasi_dari_df(df):
    hasil = []
    if df is None or df.empty: return hasil
    for _, row in df.iterrows():
        nama = row['Nama']; lat = row['Latitude']; lon = row['Longitude']
        tipe = str(row['Tipe']).strip()
        desk = row['Deskripsi'] if pd.notna(row['Deskripsi']) else ''
        if tipe in ('Wisata Sejarah','Wisata Alam'): obj = TempatWisata(nama,lat,lon,desk)
        elif tipe == 'Kuliner': obj = Kuliner(nama,lat,lon,desk)
        elif tipe == 'Tempat Ibadah': obj = TempatIbadah(nama,lat,lon,desk)
        elif tipe == 'Landmark': obj = Landmark(nama,lat,lon,desk)
        else: obj = TempatWisata(nama,lat,lon,desk)
        hasil.append(obj)
    return hasil
print('--- Praktikum 04 ---')
list_semua_lokasi = buat_objek_lokasi_dari_df(df_lokasi)
print(f'{len(list_semua_lokasi)} objek dibuat:')
for i, l in enumerate(list_semua_lokasi): print(f'  {i+1}. {repr(l)}')
```

## Praktikum 05: Visualisasi Peta dengan Folium
Marker warna beda tiap tipe (isinstance). Popup pake get_info_popup() (polymorphism).

```python
import folium
NAMA_FILE_PETA = 'peta_interaktif_semarang.html'
PUSAT = [-6.9929, 110.4200]
ZOOM = 13
def buat_peta_lokasi_folium(list_lokasi, file_output):
    if not list_lokasi: print('list kosong'); return None
    peta = folium.Map(location=PUSAT, zoom_start=ZOOM, tiles='OpenStreetMap')
    ok = 0
    for l in list_lokasi:
        lat, lon = l.get_koordinat()
        if lat is None: continue
        if isinstance(l, TempatWisata): w, ik = 'blue', 'info-sign'
        elif isinstance(l, Kuliner): w, ik = 'red', 'cutlery'
        elif isinstance(l, TempatIbadah): w, ik = 'green', 'star'
        elif isinstance(l, Landmark): w, ik = 'orange', 'flag'
        else: w, ik = 'gray', 'map-marker'
        folium.Marker([lat,lon], popup=folium.Popup(l.get_info_popup(),max_width=300),
            tooltip=l.nama, icon=folium.Icon(color=w,icon=ik,prefix='glyphicon')).add_to(peta)
        ok += 1
    peta.save(file_output)
    print(f'peta disimpan ({ok} marker)')
    return peta
print('--- Praktikum 05 ---')
peta_obj = buat_peta_lokasi_folium(list_semua_lokasi, NAMA_FILE_PETA)
print('buka file HTML di browser kalo peta gak muncul')
peta_obj
```

## Praktikum 06: File Handling (Menyimpan Log)
Mode 'a' (append) untuk mencatat log pembuatan peta.

```python
import datetime
FILE_LOG = 'proses_peta.log'
def tulis_log(pesan, file_log=FILE_LOG):
    t = datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S')
    with open(file_log, 'a', encoding='utf-8') as f:
        f.write(f'[{t}] {pesan}' + chr(10))
def buat_peta_dengan_log(list_lokasi, nama_file_peta):
    tulis_log(f'MULAI: {nama_file_peta} | {len(list_lokasi)} lokasi')
    if not list_lokasi: tulis_log('GAGAL: list kosong'); return None
    peta = folium.Map(location=PUSAT, zoom_start=ZOOM)
    ok = 0
    for l in list_lokasi:
        lat, lon = l.get_koordinat()
        if lat is None: tulis_log(f'SKIP: {l.nama}'); continue
        if isinstance(l,TempatWisata): w,ik = 'blue','info-sign'
        elif isinstance(l,Kuliner): w,ik = 'red','cutlery'
        elif isinstance(l,TempatIbadah): w,ik = 'green','star'
        elif isinstance(l,Landmark): w,ik = 'orange','flag'
        else: w,ik = 'gray','map-marker'
        folium.Marker([lat,lon],popup=folium.Popup(l.get_info_popup(),max_width=300),
            tooltip=l.nama,icon=folium.Icon(color=w,icon=ik,prefix='glyphicon')).add_to(peta)
        ok += 1
    try:
        peta.save(nama_file_peta)
        tulis_log(f'SUKSES: {nama_file_peta} | {ok} marker')
        print(f'peta ke-2 disimpan ({ok} marker)')
        return peta
    except Exception as e:
        tulis_log(f'ERROR: {e}'); print(f'ERROR: {e}'); return None
print('--- Praktikum 06 ---')
NAMA_FILE_PETA2 = 'peta_kedua.html'
buat_peta_dengan_log(list_semua_lokasi, NAMA_FILE_PETA2)
print()
print('isi file log:')
with open(FILE_LOG) as f: print(f.read())
```

## Tugas 2: Baca Konfigurasi dari File
Baca koordinat pusat peta dan zoom level dari config_peta.txt. Jika error, pakai default.

```python
config_text = """-6.9929
110.4200
13
"""
with open('config_peta.txt', 'w') as f: f.write(config_text)
print('config_peta.txt dibuat!')
```

```python
def baca_konfigurasi(nama_file='config_peta.txt'):
    try:
        with open(nama_file, 'r') as f:
            baris = f.readlines()
        lat = float(baris[0].strip())
        lon = float(baris[1].strip())
        zoom = int(baris[2].strip())
        print(f'config: lat={lat}, lon={lon}, zoom={zoom}')
        return lat, lon, zoom
    except (FileNotFoundError, IndexError, ValueError) as e:
        print(f'error: {e}, pake default Semarang')
        return -6.9929, 110.4200, 13
LAT, LON, ZOOM = baca_konfigurasi()
print(f'pusat peta: ({LAT}, {LON}) zoom={ZOOM}')
```

## Tugas 3: Peta Final dengan Semua Fitur
Gabungan: marker warna beda, kelas Museum, config dari file, logging.

```python
import folium, datetime
NAMA_FILE_FINAL = 'peta_final_semarang.html'
FILE_LOG = 'proses_peta.log'
def tulis_log(pesan):
    with open(FILE_LOG, 'a', encoding='utf-8') as f:
        f.write(f'[{datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")}] {pesan}' + chr(10))
def buat_peta_final(list_lokasi, file_output):
    tulis_log(f'MULAI: {file_output} | {len(list_lokasi)} lokasi')
    if not list_lokasi: tulis_log('GAGAL: list kosong'); return None
    lat, lon, zoom = baca_konfigurasi()
    peta = folium.Map(location=[lat, lon], zoom_start=zoom, tiles='OpenStreetMap')
    ok = 0
    for l in list_lokasi:
        lt, ln = l.get_koordinat()
        if lt is None: continue
        if isinstance(l, TempatWisata): w, ik = 'blue', 'info-sign'
        elif isinstance(l, Kuliner): w, ik = 'red', 'cutlery'
        elif isinstance(l, TempatIbadah): w, ik = 'green', 'star'
        elif isinstance(l, Landmark): w, ik = 'orange', 'flag'
        elif isinstance(l, Museum): w, ik = 'purple', 'education'
        else: w, ik = 'gray', 'map-marker'
        folium.Marker([lt,ln],popup=folium.Popup(l.get_info_popup(),max_width=300),
            tooltip=l.nama,icon=folium.Icon(color=w,icon=ik,prefix='glyphicon')).add_to(peta)
        ok += 1
    peta.save(file_output)
    tulis_log(f'SUKSES: {file_output} | {ok} marker')
    print(f'PETA FINAL: {file_output} ({ok} marker, pusat={lat},{lon}, zoom={zoom})')
    print('warna: biru=Wisata, merah=Kuliner, hijau=Ibadah, oranye=Landmark, ungu=Museum')
    return peta
print('--- MEMBUAT PETA FINAL ---')
peta_final = buat_peta_final(list_semua_lokasi, NAMA_FILE_FINAL)
peta_final
```
